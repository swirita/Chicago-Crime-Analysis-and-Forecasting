## Full Workflow

```python
chicago_full = df #pd.read_csv(RAW_FILE)
chicago_full
```

```python
# this cell can take up to 1 min to run
date_format = "%m/%d/%Y %I:%M:%S %p"

chicago_full['Datetime'] = pd.to_datetime(chicago_full['Date'], format=date_format)
chicago_full = chicago_full.sort_values('Datetime')
chicago_full
```

### Separate the Full Dataset by Years

```python
chicago_full['Datetime'].dt.year
```

```python
# Removing 2027
filter_pre_2027 =  chicago_full['Datetime'].dt.year<2027
chicago_2027 = chicago_full[~filter_pre_2027]
chicago_full = chicago_full[filter_pre_2027]
chicago_full
```

```python
# save the years for every crime
chicago_full["Year"] = chicago_full['Datetime'].dt.year.astype(str)
chicago_full["Year"].value_counts().sort_index()
```

```python
## Dropping unneeded columns to reduce file size
drop_cols = ["X Coordinate","Y Coordinate", "Community Area","FBI Code",
             "Case Number","Updated On",'Block','Location','IUCR']
```

```python
# save final df
chicago_final = chicago_full.drop(columns=drop_cols)
chicago_final = chicago_final.set_index('Datetime')
chicago_final
```

```python
# unique # of year bins
year_bins = chicago_final['Year'].astype(str).unique()
year_bins
```

```python
FINAL_DROP = ['Year']
```

### Set the Google Drive Save Location

```python
from google.colab import drive
drive.mount("/content/drive")
```

```python
import os
import glob

PROJECT_FOLDER = (
    "/content/drive/MyDrive/"
    "Chicago-Crime-Analysis-and-Forecasting"
)

OUTPUT_FOLDER = os.path.join(
    PROJECT_FOLDER,
    "data",
    "raw"
)

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

print("Saving yearly files to:", OUTPUT_FOLDER)
```

```python
print(f"[i] Saving CSV files to {OUTPUT_FOLDER}")

for year in year_bins:

    # Select one year
    temp_df = chicago_final.loc[year].copy()

    # Sort by datetime
    temp_df = temp_df.sort_index()

    # Remove Datetime index and Year column
    temp_df = temp_df.reset_index(drop=True)
    temp_df = temp_df.drop(columns=FINAL_DROP)

    # Create the output filename
    fname_temp = os.path.join(
        OUTPUT_FOLDER,
        f"Chicago-Crime_{year}.csv"
    )

    # Save the yearly CSV
    temp_df.to_csv(fname_temp, index=False)

    print(f"- Successfully saved {fname_temp}")
```

```python
saved_files = sorted(
    glob.glob(
        os.path.join(
            OUTPUT_FOLDER,
            "Chicago-Crime_*.csv"
        )
    )
)

print(f"{len(saved_files)} yearly CSV files were found.")

saved_files
```

```python
readme = """# Chicago Crime Data

Source:
https://data.cityofchicago.org/Public-Safety/Crimes-2001-to-Present/ijzp-q8t2

Preparation:
- Downloaded on September 3, 2026.
- Filtered to include records before January 1, 2027.
- Includes available crime records from 2001 through September 3, 2026.
- The 2026 data represents a partial year.
- Files are split into one CSV file per year.
- The full source CSV is too large for GitHub.

Example usage:

    import glob
    import pandas as pd

    folder = "data/raw/"
    crime_files = sorted(
        glob.glob(folder + "Chicago-Crime_*.csv")
    )

    df = pd.concat(
        [pd.read_csv(file) for file in crime_files],
        ignore_index=True
    )
"""

print(readme)
```

```python
readme_fpath = os.path.join(
    OUTPUT_FOLDER,
    "README.txt"
)

with open(readme_fpath, "w", encoding="utf-8") as file:
    file.write(readme)

print(f"Successfully saved {readme_fpath}")
```

### Making a Zip File

```python
files_to_zip = saved_files + [readme_fpath]

files_to_zip
```

```python
import zipfile

ZIP_FILE = os.path.join(
    PROJECT_FOLDER,
    "data",
    "Chicago_Crime_2001-2026.zip"
)

with zipfile.ZipFile(
    ZIP_FILE,
    mode="w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=9
) as zip_file:

    for file in files_to_zip:
        zip_file.write(
            file,
            arcname="Data/" + os.path.basename(file)
        )

print(f"Successfully created: {ZIP_FILE}")
```
